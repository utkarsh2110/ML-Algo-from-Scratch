# Linear Regression

## Analytical

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])
```

```python
x_avg = np.mean(x)
y_avg = np.mean(y)
```

w = cov(X, Y) / Var(X)

```python
xy_cov = (np.mean(x*y) - x_avg * y_avg)
x_var = np.var(x)
w = xy_cov / x_var
```

b = y_avg - w * x_avg

```python
b = y_avg - w * x_avg
```

```python
y_pred = w * x + b
mse = np.mean((y_pred - y) ** 2)

print("w =", w)
print("b =", b)
print("Predictions =", y_pred)
print("MSE =", mse)
```

```python
import numpy as np
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])

alp = 0.01 #learning rate
w = 0
b = 0

n = len(x)
epochs = 1000
for i in range(epochs):
  derL_w = (2/n) * np.sum((w*x + b - y) * x)
  derL_b = (2/n) * np.sum((w*x + b - y))
  w = w - alp * derL_w
  b = b - alp * derL_b
```

```python
y_pred = w * x + b
mse = np.mean((y_pred - y) ** 2)

print("w =", w)
print("b =", b)
print("Predictions =", y_pred)
print("MSE =", mse)
```

```python
import matplotlib.pyplot as plt

plt.scatter(x, y)
plt.plot(x, y_pred)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression from Scratch")
plt.show()
```
