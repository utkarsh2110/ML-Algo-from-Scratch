# KNN

```python
import numpy as np

x_new = int(input("Enter a number to be classified: "))

x = np.array([1, 2, 3, 6, 7, 8])
y = np.array([0, 0, 0, 1, 1, 1])

dist = []

for i in range(len(x)):
    dist.append((abs(x_new - x[i]), i))

dist.sort()

k = 3

f = dict()

for i in range(k):
    cls = y[dist[i][1]]

    if cls not in f:
        f[cls] = 0

    f[cls] += 1

max_value = [-1, -1]

for cls, freq in f.items():
    if freq > max_value[1]:
        max_value[0] = cls
        max_value[1] = freq

print("Predicted class:", max_value[0])
```
